## Set your API key

`llmranker` runs on top of [LiteLLM](https://docs.litellm.ai/docs/providers), so it works with OpenAI, Gemini, Anthropic, Azure, Bedrock, local Ollama models, or any of the 100+ providers LiteLLM supports. A few common ones are set up below (Gemini, OpenAI, Anthropic, Nebius) — paste a key for whichever one you want to use and leave the rest blank. The default below uses Gemini since it has a free tier, so you can run this notebook without a paid key: get one at [aistudio.google.com/apikey](https://aistudio.google.com/apikey).

```python
import os

os.environ["GEMINI_API_KEY"] = ""  # paste your key here
os.environ["OPENAI_API_KEY"] = ""  # paste your key here
os.environ["ANTHROPIC_API_KEY"] = ""  # paste your key here
os.environ["NEBIUS_API_KEY"] = ""  # paste your key here
```

## Rank candidates

```python
from llmranker import Candidate, LLMConfig, SetwiseRanker

# pick the provider you set a key for above (only one ranker line should be active)
ranker = SetwiseRanker(LLMConfig(model="gemini/gemini-flash-lite-latest"), num_child=4, k=5)
# ranker = SetwiseRanker(LLMConfig(model="gpt-4o-mini"), num_child=4, k=5)
# ranker = SetwiseRanker(LLMConfig(model="claude-3-5-sonnet-20241022"), num_child=4, k=5)
# ranker = SetwiseRanker(LLMConfig(model="nebius/google/gemma-3-27b-it"), num_child=4, k=5)

candidates = [
    Candidate(id="1", text="A budget hostel in the city center."),
    Candidate(id="2", text="A five-star beachfront resort with a spa."),
    Candidate(id="3", text="A family-run guesthouse near the old town, kid-friendly."),
]

result = ranker.rank(
    query="affordable, family friendly, near historical sites", candidates=candidates
)
print([c.id for c in result])  # ['3', '1', '2']
```
